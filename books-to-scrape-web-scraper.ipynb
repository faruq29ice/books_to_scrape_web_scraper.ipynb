### Adding Ratings to the Dataset

```python
## Creating a mapping Dictionary
rating_map= {
            "One": 1,
            "Two":2,
            "Three":3,
            "Four": 4,
            "Five": 5
            }
```

```python
## Adding rating to book data
book_data =[]
for book in books:
    title= book.find("h3").find("a")["title"]
    price= book.find("p",class_="price_color").text
    availability= book.find("p", class_= "instock availability").text.strip()

    # Extract the word
    rating = book.find("p", class_= "star-rating")["class"][1]
    rating= rating_map[rating]

    # Convert it to number
    book_info ={
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating
    }

    book_data.append(book_info)
print(len(book_data))
```

```python
## Building the Dictionary Automatically
product_info= {}
for row in rows:
    key =row.find("th").text
    value = row.find("td").text

    product_info[key]= value
product_info
```

```python
## instead of writting the same code repeatedly, i'll package it into a reuseable function
## Creating the fuction

def get_book_details(book):
    link =book.find("h3").find("a")["href"]
    full_link= urljoin(url, link)
    
    response = requests.get(full_link)
    response.encoding= response.apparent_encoding
    soup = BeautifulSoup(response.text, "html.parser")
    rows = soup.find("table").find_all("tr")

    product_info ={}

    for row in rows:
        key= row.find("th").text
        value = row.find("td").text
        product_info[key] = value
     # Dictionary containing product information   
    return product_info
```

```python
## Testing the Functions
details = get_book_details(books[0])
details
```

```python
## Combining all the listing-page Data with the Detail-page data
all_books = []
for book in books:

    title= book.find("h3").find("a")["title"]
    price= book.find("p",class_="price_color").text
    availability= book.find("p", class_= "instock availability").text.strip()

    # Extract the word
    rating = book.find("p", class_= "star-rating")["class"][1]
    rating= rating_map[rating]

    details = get_book_details(book)
    # Convert it to number
    book_info ={
        "Title": title,
        "Price": price,
        "Availability": availability,
        "Rating": rating,
        **details}
    all_books.append(book_info)
print(len(all_books))
```

```python
## Storing the Data and then print the top 5
df= pd.DataFrame(all_books)
df.head()
```
